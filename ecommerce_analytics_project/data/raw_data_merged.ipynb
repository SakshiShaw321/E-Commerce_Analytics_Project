***Amazon_dataset-Data Wrangling***

```python
#creating dataframes
df_amazon=pd.read_csv('amazon_women_shirts_full.csv')
print("Total number of rows in amazon_women_shirts_full.csv: ", len(df_amazon))
```

```python
df_amazon.shape
```

```python
df_amazon.head(2)
```

```python
df_amazon.isnull().sum()
```

```python
#Price-related repairs
df_amazon["original_price"] = df_amazon["original_price"].fillna(df_amazon["price"])
df_amazon["discount_percent"] = pd.to_numeric(df_amazon["discount_percent"], errors="coerce")
mask = df_amazon["discount_percent"].isna() & df_amazon["original_price"].gt(0)
df_amazon.loc[mask, "discount_percent"] = (
    (df_amazon["original_price"] - df_amazon["price"]) / df_amazon["original_price"] * 100
).round(2)
df_amazon["discount_percent"] = df_amazon["discount_percent"].fillna(0)
```

```python
# Engagement metrics
df_amazon["rating"] = df_amazon["rating"].fillna(0)
df_amazon["review_count"] = df_amazon["review_count"].fillna(0)
```

```python
#  Categorical / text
for col in ["seller", "size_options", "fit_type", "fabric", "pattern",
            "color", "sleeve_type", "description", "product_image"]:
    df_amazon[col] = df_amazon[col].fillna("unknown")
```

***Myntra_Dataset- Data Wrangling***

```python
#creating dataframes
df_myntra=pd.read_csv('myntra_women_shirts_full.csv')
print("Total number of rows in myntra_women_shirts_full.csv: ", len(df_myntra))
```

```python
df_myntra.head(2)
```

```python
df_myntra.shape
```

```python
df_myntra.isnull().sum()
```

```python
# If discount_percent is null, calculate it from price and original_price
df_myntra['price'] = pd.to_numeric(df_myntra['price'], errors='coerce')
df_myntra['original_price'] = pd.to_numeric(df_myntra['original_price'], errors='coerce')
df_myntra['discount_percent'] = pd.to_numeric(df_myntra['discount_percent'], errors='coerce')

calculated_discount = (
    (df_myntra['original_price'] - df_myntra['price'])
    / df_myntra['original_price']
) * 100

# Only fill rows where discount_percent is null
df_myntra['discount_percent'] = df_myntra['discount_percent'].fillna(calculated_discount)
df_myntra['discount_percent'] = df_myntra['discount_percent'].round().astype('Int64')
```

```python
#replacing the / with ,
df_myntra['size_options'].replace('/',',',inplace=True)
df_myntra['size_options']=df_myntra['size_options'].str.split(',')
df_myntra['size_options']=df_myntra['size_options'].fillna("unknown")
```

```python
import ast

def to_csv_string(x):
    if isinstance(x, list):
        return ','.join(x)
    if isinstance(x, str) and x.startswith('['):
        return ','.join(ast.literal_eval(x))   # "['S', 'M']" -> "S,M"
    return x

df_myntra['size_options'] = df_myntra['size_options'].apply(to_csv_string)
```

```python
#transforming the column names in lower case
df_myntra.columns=df_myntra.columns.str.lower()
```

```python
#checking the datatypes
df_myntra.dtypes
```
