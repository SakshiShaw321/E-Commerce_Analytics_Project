```python
#importing the libraries
import pandas as pd
import numpy as np
```

```python
#loading the dataset
df=pd.read_excel("combined_women_shirts.xlsx")
```

```python
#displaying the rows,columns in the dataset
df.shape
```

```python
#displaying the first few rows of the dataset
df.head()
```

```python
#checking for null values
df.isnull().sum()
```

```python
df["product_name"]=df["product_name"].str.replace("|",",")
```

```python
mapping={"boxy": "boxy fit",
   "regular": "regular fit", 
   "slim": "slim fit", 
   "oversized": "oversized fit",
   "relaxed": "relaxed fit", 
   "straight": "straight fit",
   "oversized": "oversize fit",
   "relaxed": "relaxed fit", 
   "straight": "straight fit"}

df["fit_type"]=df["fit_type"].str.strip().str.lower().replace(mapping)
```

```python
mapping={"full sleeve": "full sleeves",
         "short sleeve": "short sleeves",
         "roll-up": "roll up sleeves",
         "roll-up sleeve": "roll up sleeves",
         "roll-up sleeves": "roll up sleeves",
         "long sleeve": "full sleeves", 
         "half sleeve": "half sleeves",
         'balloon sleeve': 'balloon sleeves',
         'raglan sleeve': 'raglan sleeves',
         'puff sleeve': 'puff sleeves',
         'bishop sleeve': 'bishop sleeves',
         '3/4 sleeve': '3/4 sleeves',
         'cuff sleeve': 'cuff sleeves',
         'long sleeve (full sleeve)': 'full sleeves',
         'cap sleeve': 'cap sleeves',
         'regular sleeve': 'regular sleeves',
         'full length sleevs': 'full sleeves',
         "long sleeves": "full sleeves",
         'three-quarter sleeves': '3/4 sleeves'
          }
df["sleeve_type"]=df["sleeve_type"].str.strip().str.lower().replace(mapping)
```

```python
mapping={'graphic print': 'graphic',
        "floral print": "floral",
}
df['printed']=df['pattern'].str.strip().replace(mapping)
```

```python
mapping={'instock': 'in stock',
         'outofstock': 'out of stock',
         }
df['availability']=df['availability'].str.strip().replace(mapping)
```

```python
df.to_excel('women_shirts.xlsx', index=False)
```
